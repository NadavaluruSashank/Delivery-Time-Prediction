# Feature Engineering

Prepare features for model training.

```python
import sys
sys.path.append("../src")
from preprocess import load_data, engineer_features, get_feature_columns

df = load_data()
df = engineer_features(df)
feature_cols = get_feature_columns(df)
print(f"Total features: {len(feature_cols)}")
df[feature_cols].head()
```

## Feature correlation with target

```python
import matplotlib.pyplot as plt
correlations = df[feature_cols + ["delivery_time_minutes"]].corr()["delivery_time_minutes"].drop("delivery_time_minutes").sort_values()
correlations.plot(kind="barh", figsize=(8,6))
plt.title("Feature Correlation with Delivery Time")
plt.tight_layout()
plt.show()
```
