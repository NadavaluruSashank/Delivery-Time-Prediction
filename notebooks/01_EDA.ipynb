# Delivery Orders EDA

Exploratory analysis of simulated delivery order data.

```python
import sys
sys.path.append("../src")
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from data_generator import generate_delivery_data

df = generate_delivery_data()
df.head()
```

## Delivery time distribution

```python
plt.figure(figsize=(8,4))
sns.histplot(df["delivery_time_minutes"], bins=40, kde=True)
plt.title("Delivery Time Distribution")
plt.xlabel("Minutes")
plt.show()
```

## Delivery time by vehicle type

```python
plt.figure(figsize=(8,4))
sns.boxplot(data=df, x="vehicle_type", y="delivery_time_minutes")
plt.title("Delivery Time by Vehicle Type")
plt.show()
```

## Distance vs delivery time

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="distance_km", y="delivery_time_minutes", hue="traffic_density", alpha=0.6)
plt.title("Distance vs Delivery Time by Traffic Density")
plt.show()
```
