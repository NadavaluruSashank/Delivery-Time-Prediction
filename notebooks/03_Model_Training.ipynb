# Model Training

Train and evaluate the XGBoost delivery time prediction model.

```python
import sys
sys.path.append("../src")
from preprocess import load_data, engineer_features, get_feature_columns
from train import train_model, get_feature_importance

df = load_data()
df = engineer_features(df)
feature_cols = get_feature_columns(df)

model, metrics = train_model(df, feature_cols)
print(metrics)
```

## Feature importance

```python
import matplotlib.pyplot as plt
importance = get_feature_importance(model, feature_cols)
importance.head(10).plot(kind="barh", x="feature", y="importance", legend=False, figsize=(8,5))
plt.title("Top 10 Feature Importance")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```
